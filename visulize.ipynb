```python
import numpy as np
import trimesh

def create_uniform_crop_mesh(
    field_width=10.0,
    field_length=10.0,
    row_spacing=0.75,
    plant_spacing=0.50,
    crop_height=0.75,
):
    vertices = []
    faces = []

    y_positions = np.arange(
        -field_length / 2,
        field_length / 2,
        plant_spacing,
    )

    x_positions = np.arange(
        -field_width / 2,
        field_width / 2,
        row_spacing,
    )

    for x in x_positions:
        for y in y_positions:
            i = len(vertices)

            half_width = 0.04
            half_depth = 0.04

            vertices.extend([
                [x - half_width, y - half_depth, 0.0],
                [x + half_width, y - half_depth, 0.0],
                [x + half_width, y + half_depth, crop_height],
                [x - half_width, y + half_depth, crop_height],
            ])

            faces.extend([
                [i, i + 1, i + 2],
                [i, i + 2, i + 3],
            ])

    return trimesh.Trimesh(
        vertices=np.asarray(vertices),
        faces=np.asarray(faces),
        process=True,
    )

mesh = create_uniform_crop_mesh()


vertices = np.asarray(mesh.vertices)
faces = np.asarray(mesh.faces)

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

ax.plot_trisurf(
    vertices[:, 0],
    vertices[:, 1],
    vertices[:, 2],
    triangles=faces,
    linewidth=0,
    alpha=0.9,
)

ax.set(
    xlabel="X",
    ylabel="Y",
    zlabel="Z",
)
ax.set_box_aspect(
    np.ptp(vertices, axis=0)
)
plt.tight_layout()
plt.show()
#mesh.export("cropland_uniform_centered.ply")
```

```python
from pathlib import Path
import numpy as np
import trimesh
import matplotlib.pyplot as plt

# Source tree meshes
DATA_DIR = Path("/home/guox/Documents/s2gos-mono/extra_data/meshes")
trunk_path = DATA_DIR / "71_mesh_centered.ply"
leaves_path = DATA_DIR / "71_leaves_centered.ply"

# Field geometry
row_count = 14
line_count = 20
row_spacing = 0.75
line_spacing = 0.50
tree_height = 2.0
height_variation = 0.10
random_rotation = True
seed = 42

preview_every = 20
export_combined_mesh = False
output_path = DATA_DIR / "cropland_tree_field.ply"

# Load one copy of each component
trunk_template = trimesh.load(trunk_path, process=False)
leaves_template = trimesh.load(leaves_path, process=False)

def prepare_tree_template(template):
    vertices = np.asarray(template.vertices, dtype=float).copy()

    # Convert Y-up coordinates to Z-up coordinates:
    # [X, Y, Z] -> [X, Z, Y]
    vertices = vertices[:, [0, 2, 1]]

    # Put the lowest vertex at ground level
    vertices[:, 2] -= vertices[:, 2].min()
    template.vertices = vertices
    return template

trunk_template = prepare_tree_template(trunk_template)
leaves_template = prepare_tree_template(leaves_template)

# Scale trunk and leaves to the same target height
trunk_height = trunk_template.extents[2]
leaves_height = leaves_template.extents[2]
source_height = max(trunk_height, leaves_height)

trunk_template.apply_scale(tree_height / source_height)
leaves_template.apply_scale(tree_height / source_height)

# Generate evenly spaced row and line positions
x_positions = (
    np.arange(row_count) - (row_count - 1) / 2.0
) * row_spacing

y_positions = (
    np.arange(line_count) - (line_count - 1) / 2.0
) * line_spacing

rng = np.random.default_rng(seed)
trunk_instances = []
leaves_instances = []
positions = []

for row_index, x in enumerate(x_positions):
    for line_index, y in enumerate(y_positions):
        trunk_instance = trunk_template.copy()
        leaves_instance = leaves_template.copy()

        if height_variation > 0:
            factor = rng.uniform(
                1.0 - height_variation,
                1.0 + height_variation,
            )
            trunk_instance.apply_scale(factor)
            leaves_instance.apply_scale(factor)

        angle = rng.uniform(0, 360) if random_rotation else 0.0
        point = np.array([x, y, 0.0])

        trunk_instance.apply_translation(point)
        leaves_instance.apply_translation(point)

        if random_rotation:
            trunk_rotation = trimesh.transformations.rotation_matrix(
                np.radians(angle),
                [0, 0, 1],
                point=point,
            )
            trunk_instance.apply_transform(trunk_rotation)
            leaves_instance.apply_transform(trunk_rotation)

        trunk_instances.append(trunk_instance)
        leaves_instances.append(leaves_instance)
        positions.append((x, y))

# Preview without duplicating all triangles
fig = plt.figure(figsize=(12, 9))
ax = fig.add_subplot(111, projection="3d")

# Tree positions
xs = np.array([position[0] for position in positions])
ys = np.array([position[1] for position in positions])
ax.scatter(xs, ys, np.zeros(len(positions)), s=8, c="tab:green")

# Preview selected trunk and leaf instances
for index in range(0, len(trunk_instances), preview_every):
    trunk = trunk_instances[index]
    leaves = leaves_instances[index]

    trunk_vertices = np.asarray(trunk.vertices)
    trunk_faces = np.asarray(trunk.faces)
    ax.plot_trisurf(
        trunk_vertices[:, 0],
        trunk_vertices[:, 1],
        trunk_vertices[:, 2],
        triangles=trunk_faces,
        color="saddlebrown",
        linewidth=0,
        alpha=0.8,
    )

    leaves_vertices = np.asarray(leaves.vertices)
    leaves_faces = np.asarray(leaves.faces)
    ax.plot_trisurf(
        leaves_vertices[:, 0],
        leaves_vertices[:, 1],
        leaves_vertices[:, 2],
        triangles=leaves_faces,
        color="forestgreen",
        linewidth=0,
        alpha=0.45,
    )

ax.set(
    xlabel="X (m)",
    ylabel="Y (m)",
    zlabel="Z (m)",
)

ax.set_box_aspect([
    max(1.0, row_spacing * row_count),
    max(1.0, line_spacing * line_count),
    tree_height,
])

ax.view_init(elev=25, azim=-60)
plt.tight_layout()
plt.show()

if export_combined_mesh:
    combined = trimesh.util.concatenate(
        trunk_instances + leaves_instances
    )
    combined.export(output_path)
    print(f"Saved combined mesh: {output_path}")
else:
    print("Combined export disabled.")

tree_count = row_count * line_count
field_area = (
    max(row_spacing * (row_count - 1), 0.1)
    * max(line_spacing * (line_count - 1), 0.1)
)

print(f"Trees: {tree_count}")
print(f"Density: {tree_count / field_area:.3f} trees/m2")
print(f"Density: {tree_count / field_area * 10000:.1f} trees/ha")
print(f"Row positions: {x_positions}")
print(f"Line positions: {y_positions}")
```
