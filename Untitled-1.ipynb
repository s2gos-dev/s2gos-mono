```python
from pathlib import Path
import matplotlib.pyplot as plt
import xarray as xr

spectra_dir = Path(
    "/home/guox/Documents/s2gos-mono/"
    "packages/s2gos-generator/resources/data/spectra"
)

names = [
    "spectrum_cropland_foliage_uniform",
    "spectrum_cropland_foliage_shrubbery",
    "spectrum_cropland_foliage_trees",
]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for name in names:
    ds = xr.open_dataset(spectra_dir / f"{name}.nc")

    if "reflectance" in ds:
        axes[0].plot(ds.w, ds.reflectance, label=name)

    if "transmittance" in ds:
        axes[1].plot(ds.w, ds.transmittance, label=name)

axes[0].set(
    title="Reflectance",
    xlabel="Wavelength (nm)",
    ylabel="Reflectance",
)

axes[1].set(
    title="Transmittance",
    xlabel="Wavelength (nm)",
    ylabel="Transmittance",
)

for ax in axes:
    ax.grid(alpha=0.3)
    ax.legend()

plt.tight_layout()
plt.show()
```
